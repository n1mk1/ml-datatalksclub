```python
# Q2. Records count
dataset = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")
print(len(dataset))
dataset.head()
```

```python
# Q6. Median value of horsepower
# Median horsepower 
median = dataset["horsepower"].median()
median
```

```python
# Most frequent horsepower i.e mode
most_frequent = dataset["horsepower"].mode()[0]
most_frequent
```

```python
# Fillna
dataset["horsepower"] = dataset["horsepower"].fillna(most_frequent)
```

```python
# Median new
median_new = dataset["horsepower"].median()
median_new
```

```python
# Q7. Sum of weights
X = dataset[dataset["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).values
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.linalg.inv(X.T @ X) @ X.T @ y
w
```
